# Case: Desempenho Estudantil — Análise Exploratória com Python

## Pergunta central

**Quais características estão associadas às diferenças de desempenho em matemática, leitura e escrita?**

O objetivo é realizar uma análise descritiva e comparativa. O dataset não permite concluir causalidade.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/students_performance.csv")
df.head()
```

## 1. Diagnóstico da base

```python
print("Dimensões:", df.shape)
print("\nValores ausentes:")
print(df.isna().sum())
print("\nDuplicatas:", df.duplicated().sum())
print("\nTipos:")
print(df.dtypes)
```

## 2. Criação de métricas derivadas

```python
score_cols = ["math score", "reading score", "writing score"]

df["average score"] = df[score_cols].mean(axis=1)
df["passed_all_60"] = (df[score_cols] >= 60).all(axis=1)

df[score_cols + ["average score", "passed_all_60"]].head()
```

## 3. Estatísticas gerais

```python
df[score_cols + ["average score"]].describe().round(2)
```

## 4. Desempenho por área

```python
df[score_cols].mean().sort_values().plot(
    kind="barh",
    figsize=(9, 5),
    title="Média por área avaliada"
)
plt.xlabel("Nota média")
plt.show()
```

## 5. Curso preparatório

```python
df.groupby("test preparation course")[
    score_cols + ["average score"]
].mean().round(2)
```

```python
df.groupby("test preparation course")["average score"].mean().plot(
    kind="bar",
    figsize=(8, 5),
    title="Nota média por participação no curso preparatório"
)
plt.ylabel("Nota média geral")
plt.xticks(rotation=0)
plt.show()
```

## 6. Tipo de almoço

```python
df.groupby("lunch")[
    score_cols + ["average score"]
].mean().round(2)
```

## 7. Escolaridade parental

```python
df.groupby("parental level of education")["average score"]\
  .mean()\
  .sort_values(ascending=False)\
  .round(2)
```

## 8. Gênero e áreas de desempenho

```python
df.groupby("gender")[score_cols].mean().round(2)
```

A comparação deve ser lida de forma **descritiva**, sem interpretar gênero como causa das diferenças observadas.

## 9. Correlação entre notas

```python
df[score_cols].corr().round(3)
```

## 10. Síntese dos achados

- A média de matemática é menor do que as médias de leitura e escrita.
- Estudantes que concluíram o curso preparatório apresentam média geral superior no dataset.
- O grupo com almoço padrão apresenta média geral superior ao grupo com almoço gratuito/reduzido.
- As médias aumentam conforme algumas categorias de escolaridade parental, com destaque para mestrado e bacharelado.
- Leitura e escrita apresentam correlação muito elevada.
- As diferenças observadas são **associações descritivas** e não demonstram causalidade.

### Limitações e ética

- As categorias `race/ethnicity` estão anonimizadas como grupos A–E. O projeto não tenta atribuir identidades raciais reais a essas categorias.
- O dataset não informa país, instituição, ano letivo, idade, renda ou desenho amostral.
- O campo `lunch` pode funcionar como marcador socioeconômico, mas não deve ser tratado como medida direta de renda.
- A análise é educacional e exploratória.
